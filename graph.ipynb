```polyglot-notebook
using Microsoft.DotNet.Interactive.Formatting;
Formatter.Register(typeof(ScottPlot.Plot), (p, w) => 
    w.Write(((ScottPlot.Plot)p).GetPngHtml(800, 600)), HtmlFormatter.MimeType);
```

```polyglot-notebook
double[] thread = new double[] {1, 2, 4, 8, 16, 32, 64};
double[] times = new double[] { 0.041, 0.049, 0.025, 0.034, 0.020, 0.025, 0.041}; 

ScottPlot.Plot plt = new ScottPlot.Plot();
plt.Add.Scatter(thread, times);
plt.XLabel("Количество потоков");
plt.YLabel("Время выполнения, мс");

plt.SavePng("data.png", 600,400);

plt
```
